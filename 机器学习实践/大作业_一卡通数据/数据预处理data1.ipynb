```python
# 转换日期格式
data2['Date'] = pd.to_datetime(data2['Date'],format='%Y/%m/%d %H:%M')
new_data2 = data2.drop(['TermSerNo', 'conOperNo'],axis=1)
new_data2["day"] = new_data2["Date"].dt.day
new_data2["weekday"] = new_data2["Date"].dt.weekday
new_data2["hour"] = new_data2["Date"].dt.hour
new_data2["minute"] = new_data2["Date"].dt.minute
```

```python
def merge_record_in_an_hour(df:pd.DataFrame):
    # 同类消费记录按卡的消费次数降序，减去上一次的消费时间得到消费间隔
    # 消费间隔小于一小时的视为同一次刷卡
    # 将同一次刷卡记录合并为一次记录
    return df.groupby(((df["Date"] -  df["Date"].shift(1)) < np.timedelta64(4, 'h')).cumsum(), as_index=False).agg(Money=("Money", "sum"),FundMoney=("FundMoney", "sum"),Surplus=("Surplus", "min"),Date=("Date", "last"))
```

```python
%%time
# 按卡号，消费类型，消费地点，消费项目的编码，找出同一个人的同类消费记录
data2_merged = new_data2.groupby(['CardNo','Type', 'TermNo', 'Dept']).apply(merge_record_in_an_hour)
```
