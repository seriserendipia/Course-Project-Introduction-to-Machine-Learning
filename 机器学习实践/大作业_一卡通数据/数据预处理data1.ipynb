```python
# 转换日期格式
data2['Date'] = pd.to_datetime(data2['Date'],format='%Y/%m/%d %H:%M')
new_data2 = data2.drop(['TermSerNo', 'conOperNo'],axis=1)
```

```python
def merge_record_in_an_hour(df:pd.DataFrame):
    # 同类消费记录按卡的消费次数降序，减去上一次的消费时间得到消费间隔
    # 消费间隔小于3分钟的视为同一次刷卡
    # 将同一次刷卡记录合并为一次记录
    return df.groupby(((df["Date"] -  df["Date"].shift(1)) < np.timedelta64(3, 'm')).cumsum(), as_index=False).agg(Money=("Money", "sum"),FundMoney=("FundMoney", "sum"),Surplus=("Surplus", "min"),Date=("Date", "last"))
```

```python
%%time
# 按卡号，消费类型，消费地点，消费项目的编码，找出同一个人的同类消费记录
data2_merged = new_data2.head(100).groupby(['CardNo','Type', 'TermNo', 'Dept']).apply(merge_record_in_an_hour)
```

```python
%%time
# 按卡号，消费类型，消费地点，消费项目的编码，找出同一个人同一时刻消费记录合并
data2_merged1 = new_data2.groupby(['CardNo', 'Type', 'TermNo', 'Dept',"Date"]).agg(
        Money=("Money", "sum"), FundMoney=("FundMoney", "sum"), Surplus=("Surplus", "min"))
```

```python
data2_merged1 = data2_merged1.reset_index()
data2_merged1
```

```python
data2_merged1["day"] = data2_merged1["Date"].dt.day
data2_merged1["weekday"] = data2_merged1["Date"].dt.weekday
data2_merged1["hour"] = data2_merged1["Date"].dt.hour
data2_merged1["minute"] = data2_merged1["Date"].dt.minute
```

```python
canteen_data = data2_merged1.loc[(data2_merged1["Type"] == "消费") & (data2_merged1["Dept"].str.contains("食堂"))].copy()
```

```python
canteen_data_breakfast = canteen_data.loc[canteen_data["hour"].isin([5,6,7,8,9])].copy().groupby("Dept")["CardNo"].count()
canteen_data_lunch = canteen_data.loc[canteen_data["hour"].isin([10,11,12,13])].copy().groupby("Dept")["CardNo"].count()
canteen_data_dinner = canteen_data.loc[canteen_data["hour"].isin([16,17,18,19,20])].copy().groupby("Dept")["CardNo"].count()
```
