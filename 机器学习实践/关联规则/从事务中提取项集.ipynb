```python
from itertools import combinations

def get_frequent_items(items_list):
    groups = []
    for k in range(1,len(items_list) + 1):
        group = combinations(items_list,k)
        for i in group:
            groups.append(i)
    # groups.sort()
    print(groups)

get_frequent_items([['i1', 'i2', 'i3'], 'i4', 'i5', 'i6'])
```
