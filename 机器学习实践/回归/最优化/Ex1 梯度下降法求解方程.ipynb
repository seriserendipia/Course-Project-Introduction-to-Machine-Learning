梯度下降法求解： y=9.7966+5.4160x1+12.5843x2

```python
x1_0 = 1
x2_0 = 1
alpha = 0.05

def x1_gradient(x1,x2):
    return 5.4160
def x2_gradient(x1,x2):
    return  12.5843
def f1(x1,x2):
    return 9.7966+5.4160*x1+12.5843*x2

x1 = x1_0
x2 = x2_0
z_current = f1(x1,x2)
z_change = z_current

#存放迭代的点坐标机对应函数值列表
GD_X1=[]
GD_X2=[]
GD_Z=[]

item_num = 0
GD_X1.append(x1)
GD_X2.append(x2)
GD_Z.append(z_current)
while z_change > 1e-10 and (item_num < 100):
    item_num = item_num + 1
    x1 = x1 - x1_gradient(x1,x2)*alpha
    x2 = x2 - x2_gradient(x1,x2)*alpha
    temp = f1(x1,x2)
    z_change = z_current - temp
    z_current = temp
    GD_X1.append(x1)
    GD_X2.append(x2)
    GD_Z.append(z_current)



print(u'最终结果：（%.5f,%.5f,%.5f)'%(x1,x2,z_current))
print('迭代次数：',item_num)
print(GD_X1)
print(GD_X2)
print(GD_Z)
```

```python
x1 = np.arange(-4,4.5,0.05)
x2 = np.arange(-4,4.5,0.05)
X1,X2 = np.meshgrid(x1,x2)
Z = np.array(list(map(lambda t:f1(t[0],t[1]),
                      zip(X1.flatten(),X2.flatten())
                      )))
Z.shape = X1.shape
```

```python
fig = plt.figure()
ax = Axes3D(fig)
ax.plot_surface(x1,x2,Z, rstride=1, cstride=1, cmap='rainbow')
ax.plot(GD_X1,GD_X2,GD_Z,'ro--')
ax.set_title('求解函数 y=9.7966+5.4160*x1+12.5843*x2')
plt.show()
```
