#### 梯度下降法求解 y=9.7966+5.4160*x1+12.5843*x2

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

#二元函数表达式
def f1(x1,x2):
    return 9.7966+5.4160*x1+12.5843*x2
#对x1求偏导
def hx1(x1,x2):
    return 5.4160
#对x2求偏导
def hx2(x1,x2):
    return 12.5843
```

```python
#初始值
x1=1
x2=1
#存放迭代的点坐标机对应函数值列表
GD_X1=[]
GD_X2=[]
GD_Z=[]
#对应函数之的变化量及真实值
z_current=f1(x1,x2)
z_change=z_current
#学习率
alpha=0.05
#迭代次数
item_num=0
GD_X1.append(x1)
GD_X2.append(x2)
GD_Z.append(z_current)
```

```python
#迭代过程
while z_change>1e-10 and item_num<100:
    item_num+=1
    #改变x1和x2的值
    x1=x1-alpha*hx1(x1,x2)
    x2=x2-alpha*hx2(x1,x2)
    tmp=z_current
    z_current=f1(x1,x2)
    z_change=np.abs(tmp-z_current)
    #存储当前情况下对应函数值
    GD_X1.append(x1)
    GD_X2.append(x2)
    GD_Z.append(z_current)
print(u'最终结果：（%.5f,%.5f,%.5f)'%(x1,x2,z_current))
print('迭代次数：',item_num)
print(GD_X1)
print(GD_X2)
print(GD_Z)
```

```python
#作图
X1=np.arange(-4,4.5,0.05)
X2=np.arange(-4,4.5,0.05)
X1,X2=np.meshgrid(X1,X2)
Z=np.array(list(map(lambda t:f1(t[0],t[1]),zip(X1.flatten(),X2.flatten()))))
Z.shape=X1.shape
fig=plt.figure(facecolor='w')
ax=Axes3D(fig)
ax.plot_surface(X1,X2,Z,rstride=1,cstride=1,cmap=plt.cm.jet)
ax.plot(GD_X1,GD_X2,GD_Z,'ro--')
ax.set_title('求解函数 y=9.7966+5.4160*x1+12.5843*x2')
plt.show()
```
