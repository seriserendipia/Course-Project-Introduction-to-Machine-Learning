```python
def like_the_same_two_items_users_list(data,item1,item2):
    return data.loc[(data[item1] == 1) & (data[item2] ==1)]
```

```python
def like_the_same_two_items_users_count(data,item1,item2):
    return len(data.loc[(data[item1] == 1) & (data[item2] ==1)])
```
