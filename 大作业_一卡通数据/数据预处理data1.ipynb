```python
new_data2 = data2[data2['TermSerNo'].isnull()].copy()
new_data2.drop(['TermSerNo', 'conOperNo'],axis=1)
new_data2
```

### 去除异常值

```python
def box_plot_outliers(s):
    print(s)
    q1, q3 = s.quantile(.25), s.quantile(.75)
    iqr = q3 - q1
    low, up = q1 - 1.5*iqr, q3 + 1.5*iqr
    outlier = s.mask(((s<low) & (s != 0)) | (s>up))
    return outlier

new_data2.loc[:,['Money']] = new_data2.loc[:, ['Money']].apply(box_plot_outliers,axis = 0).copy()
new_data2 = pd.DataFrame(new_data2).dropna(subset=['Money'])
```

```python
new_data2
```

```python
%%time
# 按卡号，消费类型，消费地点，消费项目的编码，找出同一个人同一时刻消费记录合并
data2_merged2 = new_data2.groupby(['CardNo', 'Type', 'TermNo', 'Dept',"Date"]).agg(
        Money=("Money", "sum"), FundMoney=("FundMoney", "sum"), Surplus=("Surplus", "min"))
```

```python
data2_merged2 = data2_merged2.reset_index()
data2_merged2
```

```python
def divide_meal(meal_time):
    if meal_time in [5,6,7,8,9]:
        return "breakfast"
    elif meal_time in [10,11,12,13]:
        return "lunch"
    elif meal_time in [16,17,18,19,20]:
        return "dinner"
    else:
        return "not_meal_time"
```

```python
data2_merged2["day"] = data2_merged2["Date"].dt.day
data2_merged2["weekday"] = data2_merged2["Date"].dt.weekday
data2_merged2["hour"] = data2_merged2["Date"].dt.hour
data2_merged2["minute"] = data2_merged2["Date"].dt.minute
data2_merged2["meal_time"] = data2_merged2["hour"].apply(divide_meal)
```

```python
data2_merged2[(data2_merged2["hour"] == 0)& (data2_merged2["minute"] == 0)][["Dept","TermNo"]].value_counts()
```
